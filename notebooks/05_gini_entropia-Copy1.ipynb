### Resultados e Interpretación para `estado_civil_id`

Al aplicar ambos indicadores sobre las frecuencias de las $6$ categorías de la variable `estado_civil_id`, se obtuvieron los siguientes valores:

#### Interpretación de los resultados:

1. **Entropía de Shannon ($2.2466$):** 
   Este valor es alto (considerando que el máximo teórico para $6$ categorías es $\approx 2.585$). Nos indica que hay una **gran uniformidad y dispersión** en la muestra; es decir, la población se encuentra ampliamente repartida entre los diferentes estados civiles en lugar de concentrarse en uno solo.

2. **Coeficiente de Gini aplicado a las frecuencias ($0.3717$):** 
   Este valor moderado-bajo de Gini complementa la lectura anterior. Al medir la concentración relativa entre las frecuencias de las categorías, un Gini de $0.3717$ confirma que **no existe una desigualdad extrema o un acaparamiento absoluto** por parte de una sola categoría. Aunque algunos estados civiles pueden tener más registros que otros, la distribución general se mantiene relativamente equilibrada y sin picos dominantes extremos.
